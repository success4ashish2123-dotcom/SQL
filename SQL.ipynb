1.  Explain the fundamental differences between DDL, DML, and DQL
commands in SQL. Provide one example for each type of command.

Answer.

DDL (Data Definition Language): DDL commands are used to create, modify, and delete the structure of a database or table. For example, the CREATE command is used to create a new table in a database. The following SQL statement creates a table named students with two columns, student_id and student_name.
CREATE TABLE students (
    student_id INT,
    student_name VARCHAR(50)
);

DML (Data Manipulation Language): DML commands are used to insert, update, and delete data stored in database tables. For example, the INSERT command adds a new record to a table.

INSERT INTO students (student_id, student_name)
VALUES (1, 'Ashish');

DQL (Data Query Language): DQL commands are used to retrieve data from a database. The SELECT command is the primary DQL command used to fetch specific information from a table. For example:
SELECT student_name
FROM students
WHERE student_id = 1;


2. What is the purpose of SQL constraints? Name and describe three common types
of constraints, providing a simple scenario where each would be useful.


Answer.

SQL constraints are rules applied to database tables to maintain the accuracy, validity, and integrity of data. They prevent invalid or duplicate data from being stored in a table. Three common types of SQL constraints are PRIMARY KEY, NOT NULL, and UNIQUE.
1. PRIMARY KEY Constraint: The PRIMARY KEY constraint uniquely identifies each record in a table. It does not allow duplicate or NULL values. For example, in a student database, every student must have a unique student ID.
CREATE TABLE students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50)
);
2. NOT NULL Constraint: The NOT NULL constraint ensures that a column cannot contain NULL values. It is useful when a particular field must always have a value. For example, every student record should include the student's name.
CREATE TABLE students (
    student_id INT,
    student_name VARCHAR(50) NOT NULL
);
3. UNIQUE Constraint: The UNIQUE constraint ensures that all values in a column, or a combination of columns, are unique. Unlike a primary key, a table can have multiple UNIQUE constraints. For example, each student should have a different email address.
CREATE TABLE students (
    student_id INT PRIMARY KEY,
    email VARCHAR(100) UNIQUE
);

3. Explain the difference between LIMIT and OFFSET clauses in SQL. How
would you use them together to retrieve the third page of results, assuming each page
has 10 records?

Answer.

In SQL, the LIMIT and OFFSET clauses are used to control the number of records returned by a query. The LIMIT clause specifies the maximum number of records to retrieve, while the OFFSET clause specifies how many records to skip before returning results.
For example, if we want to retrieve 10 records per page, we can use LIMIT 10. The OFFSET value determines which page of records we want to display. For the first page, we use OFFSET 0; for the second page, we use OFFSET 10; and for the third page, we use OFFSET 20.
To retrieve the third page of results, assuming each page contains 10 records, we can write the following SQL query:
SELECT *
FROM students
ORDER BY student_id
LIMIT 10 OFFSET 20;

4.  What is a Common Table Expression (CTE) in SQL, and what are its main
benefits? Provide a simple SQL example demonstrating its usage.

Answer.

A Common Table Expression (CTE) in SQL is a temporary named result set that can be used within a single SQL statement. It is defined using the WITH keyword and helps simplify complex queries by breaking them into smaller, more readable parts.
The main benefits of a CTE are that it improves query readability, makes complex SQL queries easier to understand and maintain, and allows the same CTE to be referenced multiple times within a single statement. CTEs can also be recursive, which is useful for working with hierarchical data, such as employee-manager relationships.
For example, suppose we have an employees table and want to find employees whose salary is greater than ₹50,000. We can use a CTE as follows:
WITH high_salary AS (
    SELECT employee_name, salary
    FROM employees
    WHERE salary > 50000
)
SELECT *
FROM high_salary;

5. Describe the concept of SQL Normalization and its primary goals. Briefly
explain the first three normal forms (1NF, 2NF, 3NF).

Answer.

SQL Normalization is the process of organizing data in a database to reduce data redundancy and improve data integrity. Its primary goals are to avoid storing the same information unnecessarily, prevent insertion, update, and deletion anomalies, and make the database easier to maintain. Normalization is achieved by dividing large tables into smaller related tables and establishing relationships between them.
First Normal Form (1NF): A table is in 1NF when each column contains atomic values, meaning each cell contains only one value, and there are no repeating groups of data. For example, if a student has two phone numbers, they should not be stored together in a single cell. Instead, the data should be organized so that each cell contains only one phone number.
Second Normal Form (2NF): A table is in 2NF when it is already in 1NF and every non-key attribute depends on the entire candidate key, not just part of a composite key. For example, consider a table containing student_id, course_id, student_name, and course_name, where the primary key consists of student_id and course_id. Since student_name depends only on student_id and course_name depends only on course_id, the table has partial dependencies. To achieve 2NF, we can separate the data into Students, Courses, and Enrollments tables.
Third Normal Form (3NF): A table is in 3NF when it is already in 2NF and has no transitive dependencies of non-key attributes on a candidate key. In simple terms, a non-key column should not depend on another non-key column. For example, if an Employees table contains employee_id, department_id, and department_name, the department name depends on department_id, not directly on employee_id. To achieve 3NF, we can move department information into a separate Departments table and link it using department_id.

7. Generate a report showing CustomerName, Email, and the
TotalNumberofOrders for each customer. Include customers who have not placed
any orders, in which case their TotalNumberofOrders should be 0. Order the results
by CustomerName.

Answer.

SELECT
    c.CustomerName,
    c.Email,
    COUNT(o.OrderID) AS TotalNumberOfOrders
FROM Customers c
LEFT JOIN Orders o
    ON c.CustomerID = o.CustomerID
GROUP BY
    c.CustomerID,
    c.CustomerName,
    c.Email
ORDER BY
    c.CustomerName;

8.  Retrieve Product Information with Category: Write a SQL query to
display the ProductName, Price, StockQuantity, and CategoryName for all
products. Order the results by CategoryName and then ProductName alphabetically.

Answer.

SELECT
    p.ProductName,
    p.Price,
    p.StockQuantity,
    c.CategoryName
FROM Products p
INNER JOIN Categories c
    ON p.CategoryID = c.CategoryID
ORDER BY
    c.CategoryName ASC,
    p.ProductName ASC;

9. Write a SQL query that uses a Common Table Expression (CTE) and a
Window Function (specifically ROW_NUMBER() or RANK()) to display the
CategoryName, ProductName, and Price for the top 2 most expensive products in
each CategoryName.

Answer.

WITH RankedProducts AS (
    SELECT
        c.CategoryName,
        p.ProductName,
        p.Price,
        ROW_NUMBER() OVER (
            PARTITION BY c.CategoryName
            ORDER BY p.Price DESC
        ) AS RowNum
    FROM Products p
    INNER JOIN Categories c
        ON p.CategoryID = c.CategoryID
)
SELECT
    CategoryName,
    ProductName,
    Price
FROM RankedProducts
WHERE RowNum <= 2
ORDER BY CategoryName, Price DESC;

10.  You are hired as a data analyst by Sakila Video Rentals, a global movie
rental company. The management team is looking to improve decision-making by
analyzing existing customer, rental, and inventory data.

1. Identify the top 5 customers based on the total amount they’ve spent. Include customer
name, email, and total amount spent.

Answer

SELECT
    c.first_name,
    c.last_name,
    c.email,
    SUM(p.amount) AS total_amount_spent
FROM customer c
INNER JOIN payment p
    ON c.customer_id = p.customer_id
GROUP BY
    c.customer_id,
    c.first_name,
    c.last_name,
    c.email
ORDER BY total_amount_spent DESC
LIMIT 5;


2.  Which 3 movie categories have the highest rental counts? Display the category name
and number of times movies from that category were rented.

Answer.

SELECT
    c.name AS category_name,
    COUNT(r.rental_id) AS rental_count
FROM category c
INNER JOIN film_category fc
    ON c.category_id = fc.category_id
INNER JOIN inventory i
    ON fc.film_id = i.film_id
INNER JOIN rental r
    ON i.inventory_id = r.inventory_id
GROUP BY c.category_id, c.name
ORDER BY rental_count DESC
LIMIT 3;

3.  Calculate how many films are available at each store and how many of those have
never been rented.

Answer.

SELECT
    i.store_id,
    COUNT(DISTINCT i.film_id) AS total_films_available,
    COUNT(DISTINCT CASE
        WHEN r.rental_id IS NULL THEN i.film_id
    END) AS films_never_rented
FROM inventory i
LEFT JOIN rental r
    ON i.inventory_id = r.inventory_id
GROUP BY i.store_id
ORDER BY i.store_id;

4.  Show the total revenue per month for the year 2023 to analyze business seasonality.

Answer.

SELECT
    MONTH(payment_date) AS month_number,
    MONTHNAME(payment_date) AS month_name,
    SUM(amount) AS total_revenue
FROM payment
WHERE payment_date >= '2023-01-01'
  AND payment_date < '2024-01-01'
GROUP BY
    MONTH(payment_date),
    MONTHNAME(payment_date)
ORDER BY month_number;

5.  Identify customers who have rented more than 10 times in the last 6 months.

Answer.

SELECT
    c.customer_id,
    c.first_name,
    c.last_name,
    c.email,
    COUNT(r.rental_id) AS total_rentals
FROM customer c
INNER JOIN rental r
    ON c.customer_id = r.customer_id
WHERE r.rental_date >= DATE_SUB(CURDATE(), INTERVAL 6 MONTH)
  AND r.rental_date <= CURDATE()
GROUP BY
    c.customer_id,
    c.first_name,
    c.last_name,
    c.email
HAVING COUNT(r.rental_id) > 10
ORDER BY total_rentals DESC;
